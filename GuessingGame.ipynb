```python
while True:
    name = input ("Please guess my name: ")
    guesses +=1

    if name == "Mary":
        print("Congratulations!")
        print("You quessed ", guesses, "times.")
        break

    else:
       answer= input("Do you want to quit (y/n)?")
    if answer == "y":
        print("Goodbye!")
        break
    elif answer == "n":
        continue
```
